```python
import numpy as np
```

```python
class Detector:
    def __init__(self, R, N, r, a):
        self.R = R
        self.N = N
        self.r = r
        self.a = a
        
        self.pmt_coords = self._generate_pmt_coordinates()

    def _generate_pmt_coordinates(self):
        #генерация с исользованием спирали Фиббоначи
        points = []
        phi = np.pi * (3. - np.sqrt(5.)) 
        
        for i in range(self.N):
            y = 1 - (i / float(self.N - 1)) * 2
            radius = np.sqrt(1 - y * y)
            theta = phi * i
            
            x = np.cos(theta) * radius
            z = np.sin(theta) * radius
            
            points.append([x * self.R, y * self.R, z * self.R])
            
        return np.array(points)

    def is_point_in_pmt(self, point):
        point = np.array(point)
        
        if not np.isclose(np.linalg.norm(point), self.R, rtol=1e-3):
            return False, None 
            
        distances = np.linalg.norm(self.pmt_coords - point, axis=1)
        
        min_dist = np.min(distances)
        closest_pmt_idx = np.argmin(distances)
        
        if min_dist <= self.r:
            return True, closest_pmt_idx
        else:
            return False, None
```

```python
det = Detector(R=2.0, N=40, r=0.1, a=0.5)
print(f"Координаты ФЭУ:\n{det.pmt_coords}\n")

test_point_1 = det.pmt_coords[0]
is_inside_1, idx_1 = det.is_point_in_pmt(test_point_1)
print(f"Тест 1 (центр ФЭУ):")
print(f"Точка: {test_point_1}")
print(f"Попала в ФЭУ: {is_inside_1} (Индекс: {idx_1})\n")


test_point_2 = det.pmt_coords[0] + np.array([0.15, 0, 0]) 
test_point_2 = test_point_2 / np.linalg.norm(test_point_2) * det.R 

is_inside_2, idx_2 = det.is_point_in_pmt(test_point_2)
print(f"Тест 2 (мертвая зона):")
print(f"Точка: {test_point_2}")
print(f"Попала в ФЭУ: {is_inside_2} (Индекс: {idx_2})")
```
